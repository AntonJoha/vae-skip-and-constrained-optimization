```python
import json
import matplotlib.pyplot as plt

# Load parsed data
with open("../rnn_training.json", "r") as f:
    data = json.load(f)

epochs = [e["epoch"] for e in data["epochs"]]
train_loss = [e["train_loss"] for e in data["epochs"]]
val_loss = [e["val_loss"] for e in data["epochs"]]

plt.figure(figsize=(12, 6))

plt.plot(
    epochs,
    train_loss,
    label="Train Loss",
    linewidth=2,
)

plt.plot(
    epochs,
    val_loss,
    label="Validation Loss",
    linewidth=2,
)

# Highlight best validation loss
best_idx = min(range(len(val_loss)), key=lambda i: val_loss[i])

plt.scatter(
    epochs[best_idx],
    val_loss[best_idx],
    color="red",
    zorder=10,
    #label="Best Val Loss " + f"{val_loss:0.4f}"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("loss_curves.png", dpi=300)
plt.show()
```
