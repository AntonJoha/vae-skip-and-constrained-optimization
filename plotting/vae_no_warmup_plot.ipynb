```python
import json
import matplotlib.pyplot as plt


folder = "../results/vae_no_warmup/"
files = ["vae_no_warmup_short.json", "vae_no_warmup_mid.json", "vae_no_warmup_long.json"]
name = {"vae_no_warmup_short.json": "Short",  "vae_no_warmup_mid.json": "Middle", "vae_no_warmup_long.json": "Long"}
datasets = {}

for file in files:
    with open(folder + file, "r") as f:
        datasets[file] = json.load(f)



for f in files:
    print(f)
    data = datasets[f][:50]
    epochs = [e["epoch"] for e in data]
    train_loss = [e["train_loss"] for e in data]
    val_loss = [e["test_nll"] for e in data]
    
    plt.figure(figsize=(12, 6))
    
    plt.plot(
        epochs,
        train_loss,
        label="Train Loss",
        linewidth=2,
    )
    
    plt.plot(
        epochs,
        val_loss,
        label="Validation Loss",
        linewidth=2,
    )
    
    # Highlight best validation loss
    best_idx = min(ranscrabbled_val_nllge(len(val_loss)), key=lambda i: val_loss[i])
    
    plt.scatter(
        epochs[best_idx],
        val_loss[best_idx],
        color="red",
        zorder=10,
        #label="Best Val Loss " + f"{val_loss:0.4f}"
    )
    
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Training and Validation Loss " + name[f])
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig("loss_curves.png", dpi=300)
    plt.show()
```
