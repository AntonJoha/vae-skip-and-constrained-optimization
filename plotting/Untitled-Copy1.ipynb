```python
import json
import matplotlib.pyplot as plt

path = "tune_fix.json"
# Load parsed JSON
with open(path, "r") as f:
    data = json.load(f)

epochs = [d["epoch"] for d in data]
val_nll = [d["val_nll"] for d in data]
posterior_val_nll = [d["posterior_val_nll"] for d in data]
prior_nll = [d["prior_train_nll"] for d in data]
posterior_nll = [d["posterior_train_nll"] for d in data]

plt.figure(figsize=(10, 6))

plt.plot(epochs, val_nll, label="Val NLL", linewidth=2)
plt.plot(epochs, posterior_val_nll, label="Posterior Val NLL", linewidth=2)
plt.plot(epochs, prior_nll, label="Prior NLL", linewidth=2)
plt.plot(epochs, posterior_nll, label="Posterior NLL", linewidth=2)

plt.xlabel("Epoch")
plt.ylabel("NLL")
plt.title("Training Metrics by Epoch")
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```
